```python
import librosa
import numpy as np
import soundfile as sf
from IPython.display import Audio
from matplotlib import pyplot as plt
```

```python
(x, SAMPLE_RATE) = sf.read("/Users/abhijain/Documents/GitHub/photo_editor/spinopel-diesel-car-driving-away-345713.mp3")
print(x)
print(x.shape)
duplicated = np.tile(x, (2, 1))
print(duplicated.shape)
```

```python
Audio(data=duplicated, rate=SAMPLE_RATE)
```

```python
duplicated[0, :] = 0
Audio(data=duplicated, rate=SAMPLE_RATE)
```

```python
duplicated[0, :] = duplicated[1, :]
duplicated[1, :] = 0
Audio(data=duplicated, rate=SAMPLE_RATE)
```

```python
space = np.linspace(0, np.pi / 2, len(x))
duplicated[1, :] = duplicated[0, :]
duplicated[0, :] = duplicated[0, :] * np.cos(space)
duplicated[1, :] = duplicated[1, :] * np.sin(space)
Audio(data=duplicated, rate=SAMPLE_RATE)
```

```python
vcar = 5
t = len(x) / SAMPLE_RATE
dist = vcar * t
x_you = dist / 2
y_you = 5
print(dist, x_you)
```

```python
time = np.linspace(0, t, len(x))
car_pos = vcar * time
past_you = np.where(car_pos > x_you, False, True).reshape(-1,)
plt.plot(time, car_pos)
```

```python
dist_from_centre = np.abs(x_you - car_pos)
plt.plot(time, dist_from_centre)
```

```python
angle = np.where(past_you, np.atan(y_you / dist_from_centre), np.pi - np.atan(y_you / dist_from_centre))
plt.plot(time, angle * 180 / np.pi)
```

```python
v_in_direction = vcar * np.cos(angle)
plt.plot(time, v_in_direction)
```

```python
doppler_speed = 343 - v_in_direction
doppler_ratio = 343 / doppler_speed
plt.plot(time, doppler_ratio)
```

```python
remapped = np.cumsum(doppler_ratio)
remapped = remapped / max(remapped) * len(x)
print(len(remapped), min(remapped), max(remapped))

plt.plot(time, remapped)
```

```python
Audio(data=duplicated * doppler_ratio, rate=SAMPLE_RATE)
```

```python
plt.plot(time, np.sin(angle))
```

```python
Audio(data=duplicated * np.sin(angle), rate=SAMPLE_RATE)
```
